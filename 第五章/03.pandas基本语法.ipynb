```python
import pandas as pd

# 构造DataFrame -- 创建数据集(学员成绩信息)
df = pd.DataFrame([
    {"姓名": "小王", "语文": 90, "数学": 80, "英语": 70},
    {"姓名": "小李", "语文": 80, "数学": 90, "英语": 80},
    {"姓名": "小张", "语文": 70, "数学": 80, "英语": 90},
    {"姓名": "小赵", "语文": 60, "数学": 70, "英语": 80},
    {"姓名": "小孙", "语文": 80, "数学": 90, "英语": 70},
    {"姓名": "小李", "语文": 80, "数学": 80, "英语": 90},
    {"姓名": "小王2", "语文": 90, "数学": 80, "英语": 80},
    {"姓名": "小王3", "语文": 65, "数学": 55, "英语": 99},
    {"姓名": "小王4", "语文": 91, "数学": 78, "英语": 82}
])

print("原始数据：")
print(df)
print("\n" + "="*50)

# 1. 基础统计信息
print("\n1. 数据基本信息：")
print(df.info())
print("\n" + "="*50)

# 2. 描述性统计
print("\n2. 描述性统计：")
print(df.describe())
print("\n" + "="*50)

# 3. 各科成绩统计
print("\n3. 各科成绩统计：")
print(f"语文平均分: {df['语文'].mean():.2f}")
print(f"数学平均分: {df['数学'].mean():.2f}")
print(f"英语平均分: {df['英语'].mean():.2f}")
print("\n" + "="*50)

# 4. 总分计算
df['总分'] = df['语文'] + df['数学'] + df['英语']
print("\n4. 添加总分列：")
print(df[['姓名', '总分']])
print("\n" + "="*50)

# 5. 平均分计算
df['平均分'] = df[['语文', '数学', '英语']].mean(axis=1)
print("\n5. 添加平均分列：")
print(df[['姓名', '平均分']])
print("\n" + "="*50)

# 6. 按总分排序
df_sorted = df.sort_values('总分', ascending=False)
print("\n6. 按总分排序（从高到低）：")
print(df_sorted[['姓名', '总分', '平均分']])
print("\n" + "="*50)

# 7. 各科最高分和最低分
print("\n7. 各科最高分：")
print(f"语文最高分: {df['语文'].max()}")
print(f"数学最高分: {df['数学'].max()}")
print(f"英语最高分: {df['英语'].max()}")

print("\n各科最低分：")
print(f"语文最低分: {df['语文'].min()}")
print(f"数学最低分: {df['数学'].min()}")
print(f"英语最低分: {df['英语'].min()}")
print("\n" + "="*50)

# 8. 成绩分布统计
print("\n8. 成绩分布统计：")
print("语文成绩分布：")
print(df['语文'].value_counts().sort_index())
print("\n数学成绩分布：")
print(df['数学'].value_counts().sort_index())
print("\n英语成绩分布：")
print(df['英语'].value_counts().sort_index())
print("\n" + "="*50)

# 9. 筛选优秀学生（总分 > 240）
df_excellent = df[df['总分'] > 240]
print("\n9. 优秀学生（总分 > 240）：")
print(df_excellent[['姓名', '总分', '平均分']])
print("\n" + "="*50)

# 10. 各科成绩统计汇总
print("\n10. 各科成绩统计汇总：")
summary = df[['语文', '数学', '英语']].agg(['mean', 'max', 'min', 'std'])
print(summary)
print("\n" + "="*50)

# 11. 按姓名分组统计（如果有重名）
print("\n11. 按姓名分组统计（处理重名）：")
df_grouped = df.groupby('姓名').agg({
    '语文': 'mean',
    '数学': 'mean', 
    '英语': 'mean',
    '总分': 'mean'
}).round(2)
print(df_grouped)
```
